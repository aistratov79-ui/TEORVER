### Пример 1.8
> В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

* **Пункт а)** По каждой номинации установлены различные призы. Каждый из вариантов распределения представляет собой комбинацию 5 фильмов из 10, отличающуюся как составом, так и их порядком по номинациям (один фильм может получить несколько призов). Число вариантов равно числу размещений с повторениями из 10 по 5:

$$\bar{A}_{10}^5 = 10^5$$

```python
10**5
```

* **Пункт б)** По каждой номинации установлены одинаковые призы (порядок следования фильмов значения не имеет). Число вариантов распределения представляет собой число сочетаний с повторениями из 10 по 5:

$$\bar{C}_{10}^5 = C_{10 + 5 - 1}^5 = C_{14}^5 = \frac{14!}{5! \cdot (14 - 5)!}$$

```python
int(sc.comb(14, 5))
```

### Пример 1.9
> Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 — по 2 раза?

Каждое семизначное число отличается порядком следования цифр ($n_1 = 3, n_2 = 2, n_3 = 2$, сумма $n = 7$), то есть представляет собой число перестановок с повторениями из 7 элементов:

$$P_7(3, 2, 2) = \frac{7!}{3! \cdot 2! \cdot 2!}$$

```python
int(sc.factorial(7) / (sc.factorial(3) * sc.factorial(2) * sc.factorial(2)))
```

### Пример 1.10
> Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к другой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

* Рассмотрим событие $A$ – получение слова «ТОР».
Общее число вариантов выбора и упорядочивания 3 карточек из 6 равно числу размещений:
$n = A_6^3 = \frac{6!}{(6 - 3)!} = 6 \cdot 5 \cdot 4 = 120$.
Благоприятствующий исход всего один ($m = 1$):

$$P(A) = \frac{m}{n} = \frac{1}{A_6^3} = \frac{1}{6 \cdot 5 \cdot 4}$$

```python
round(1 / (sc.factorial(6) / sc.factorial(3)), 5)
```

* Рассмотрим событие $B$ – получение слова «ТЕОРИЯ».
Общее число вариантов упорядочивания всех 6 карточек равно числу перестановок:
$n = P_6 = 6! = 720$.
Благоприятствующий исход всего один ($m = 1$):

$$P(B) = \frac{m}{n} = \frac{1}{P_6} = \frac{1}{6!}$$

```python
round(1 / sc.factorial(6), 6)
```

### Пример 1.11
> Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

Рассмотрим событие $B$ – получение слова «АНАНАС».
Общее число исходов перестановки 6 карточек: $n = P_6 = 6!$.
Число благоприятствующих исходов $m$ учитывает, что перестановки одинаковых букв (3 букв «А» способами $P_3 = 3!$ и 2 букв «Н» способами $P_2 = 2!$) не меняют собранное слово:

$$m = P_3 \cdot P_2 = 3! \cdot 2!$$

$$P(B) = \frac{m}{n} = \frac{P_3 \cdot P_2}{P_6} = \frac{3! \cdot 2!}{6!}$$

```python
round((sc.factorial(3) * sc.factorial(2)) / sc.factorial(6), 4)
```

### Пример 1.16
> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

Общее число способов рассадки 10 человек на 10 пронумерованных местах: $n = P_{10} = 10!$.

* Рассмотрим событие $A$ – никакие два лица одного пола не сядут рядом.
Мужчины и женщины должны чередоваться. Если женщины занимают четные места ($5!$ способов), то мужчины занимают нечетные места ($5!$ способов), либо наоборот:

$$m_1 = 2 \cdot (5!)^2$$

$$P(A) = \frac{m_1}{n} = \frac{2 \cdot (5!)^2}{10!}$$

```python
round(2 * (sc.factorial(5) ** 2) / sc.factorial(10), 5)
```

* Рассмотрим событие $B$ – пять супружеских пар сядут рядом.
Число перестановок 5 пар равно $5!$, при этом в каждой паре муж и жена могут поменяться местами 2 способами ($2^5$ способов для 5 пар):

$$m_2 = 5! \cdot 2^5$$

$$P(B) = \frac{m_2}{n} = \frac{5! \cdot 2^5}{10!}$$

```python
round(sc.factorial(5) * (2**5) / sc.factorial(10), 5)
```

### Пример 1.17
> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

Всего в вагоне $9 \cdot 4 = 36$ мест. Общее число способов распределить 7 билетов:

$$n = C_{36}^7$$

* Рассмотрим событие $A$ – пассажиры попали в два купе.
Выбрать 2 купе из 9 можно $C_9^2$ способами, а выбрать 7 мест из имеющихся в двух купе 8 мест — $C_8^7$ способами:

$$P(A) = \frac{m_1}{n} = \frac{C_9^2 \cdot C_8^7}{C_{36}^7}$$

```python
round((sc.comb(9, 2) * sc.comb(8, 7)) / sc.comb(36, 7), 7)
```

* Рассмотрим событие $B$ – пассажиры попали в семь разных купе.
Выбрать 7 купе из 9 можно $C_9^7$ способами. В каждом выбранном купе пассажир занимает одно из 4 мест (всего $4^7$ способов):

$$P(B) = \frac{m_2}{n} = \frac{C_9^7 \cdot 4^7}{C_{36}^7}$$

```python
round((sc.comb(9, 7) * (4**7)) / sc.comb(36, 7), 5)
```

* Рассмотрим событие $D$ – пассажиры попали в три купе.
3 купе из 9 выбираются $C_9^3$ способами. Размещение 7 человек по 3 купе (вместимость купе — не более 4 человек) возможно по следующим схемам:
1. $4 + 2 + 1$: перестановки купе дают $3! = 6$ вариантов, выбор мест — $C_4^4 \cdot C_4^2 \cdot C_4^1$;
2. $3 + 3 + 1$: перестановки купе дают $\frac{3!}{2!} = 3$ варианта, выбор мест — $C_4^3 \cdot C_4^3 \cdot C_4^1$;
3. $3 + 2 + 2$: перестановки купе дают $\frac{3!}{2!} = 3$ варианта, выбор мест — $C_4^3 \cdot C_4^2 \cdot C_4^2$.

$$m_3 = C_9^3 \cdot \left(6 \cdot C_4^4 C_4^2 C_4^1 + 3 \cdot C_4^3 C_4^3 C_4^1 + 3 \cdot C_4^3 C_4^2 C_4^2\right)$$

$$P(D) = \frac{m_3}{n} = \frac{C_9^3 \cdot \left(6 \cdot C_4^4 C_4^2 C_4^1 + 3 \cdot C_4^3 C_4^3 C_4^1 + 3 \cdot C_4^3 C_4^2 C_4^2\right)}{C_{36}^7}$$

```python
comb_4_2_1 = 6 * sc.comb(4, 4) * sc.comb(4, 2) * sc.comb(4, 1)
comb_3_3_1 = 3 * sc.comb(4, 3) * sc.comb(4, 3) * sc.comb(4, 1)
comb_3_2_2 = 3 * sc.comb(4, 3) * sc.comb(4, 2) * sc.comb(4, 2)

m3 = sc.comb(9, 3) * (comb_4_2_1 + comb_3_3_1 + comb_3_2_2)
n = sc.comb(36, 7)

round(m3 / n, 5)
```

### Пример 1.48

>В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения ll очков; б) выигрыша.

```python
combinat = (list(product(range(1, 7), repeat=3)))
n = len(combinat)
a = len([elem for elem in combinat if sum(elem) == 11])
p_A = a / n
b = len([elem for elem in combinat if sum(elem) > 10])
p_B = b / n
```

* Рассмотрим вероятность события $A$ – выпало ll очков


```python
print("Вероятность события A равна:", round(p_A, 4))
```

* Рассмотрим вероятность события $B$ – мы выиграли

```python
print("Вероятность события B равна:", round(p_B, 4))
```
